```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
class SuperstoreAnalysis:
    def __init__(self, file_path):
        self.file_path = file_path
        self.df = None
        
    def load_data(self, sheet_name='Orders'):
        """Load dataset from Excel file"""
        try:
            self.df = pd.read_excel(self.file_path, sheet_name=sheet_name)
            print(f"Dataset '{sheet_name}' loaded successfully. Total rows: {len(self.df)}")
            return self.df
        except Exception as e:
            print(f"Failed to load dataset: {e}")
            return None

# # ---------------   clean_and_engineer_features   ---------------------------
    
    def clean_and_engineer_features(self):
        """Perform data cleaning and feature engineering"""
        if self.df is None:
            print("No data loaded. Cannot perform cleaning")
            return
        
        # Remove duplicates
        initial_rows = len(self.df)
        self.df.drop_duplicates(inplace=True)
        print(f"Cleaned data: Removed {initial_rows - len(self.df)} duplicate rows")
        
        # Convert date columns
        self.df['Order Date'] = pd.to_datetime(self.df['Order Date'])
        self.df['Ship Date'] = pd.to_datetime(self.df['Ship Date'])
        
        # Create Shipping Duration
        self.df['Shipping Duration'] = (self.df['Ship Date'] - self.df['Order Date']).dt.days
        
        # Calculate Profit Margin
        self.df['Profit Margin'] = np.where(
            self.df['Sales'] != 0, 
            (self.df['Profit'] / self.df['Sales']) * 100, 
            0
        )
        
        # Create Sales Performance Category
        sales_bins = [-np.inf, 50, 250, np.inf]
        sales_labels = ['Low', 'Medium', 'High']
        self.df['Sales Performance Category'] = pd.cut(
            self.df['Sales'], 
            bins=sales_bins, 
            labels=sales_labels
        )
        print("Data cleaning and feature engineering completed")


# ---------------   optimize_memory    ---------------------------    

    def optimize_memory_usage(self):
        """Optimize dataframe memory usage by converting object columns to categories"""
        if self.df is None: return
        start_mem = self.df.memory_usage().sum() / 1024**2
        for col in self.df.select_dtypes(include=['object']).columns:
            if self.df[col].nunique() / len(self.df) < 0.5:
                self.df[col] = self.df[col].astype('category')
        end_mem = self.df.memory_usage().sum() / 1024**2
        print(f"Memory optimization: Reduced from {start_mem:.2f} MB to {end_mem:.2f} MB.")

# ---------------   Handle Outliers   ---------------------------

    def handle_outliers(self, column='Sales'):
        """Detect and handle outliers in numerical columns using the IQR method"""
        if self.df is None:
            print("No data available for outlier handling")
            return
        
        Q1 = self.df[column].quantile(0.25)
        Q3 = self.df[column].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        outliers_count = ((self.df[column] < lower_bound) | (self.df[column] > upper_bound)).sum()
        print(f"Detected {outliers_count} outliers in column '{column}'.")
        
        self.df[column] = np.where(self.df[column] > upper_bound, upper_bound, self.df[column])
        self.df[column] = np.where(self.df[column] < lower_bound, lower_bound, self.df[column])
        print(f"Outliers in '{column}' have been successfully capped.")

    def export_cleaned_data(self, output_file_path='Cleaned_Superstore.csv'):
        """Automatically export the cleaned dataset to a CSV file"""
        if self.df is None:
            print("No data available to export")
            return
        
        try:
            self.df.to_csv(output_file_path, index=False)
            print(f"Cleaned dataset successfully exported to '{output_file_path}'")
        except Exception as e:
            print(f"Failed to export dataset: {e}")


    

# ---------------   EDA   ---------------------------
    
    def perform_eda_checks(self):
        """Perform additional EDA checks including missing values and unique counts"""
        if self.df is None:
            print("No data available for EDA checks")
            return
        
        print("--- Missing Values Check ---")
        missing_values = self.df.isnull().sum()
        if missing_values.sum() > 0:
            display(missing_values[missing_values > 0])
        else:
            print("Great! No missing values found in the dataset")
        
        print("\n--- Categorical Summary (Unique Values) ---")
        cat_cols = self.df.select_dtypes(include=['object', 'category'])
        for col in cat_cols.columns:
            print(f"Column '{col}' has {self.df[col].nunique()} unique values.")

    
    
    def get_statistical_summary(self):
        """Generate statistical summary and correlation matrix for numerical features"""
        if self.df is None:
            print("No data available for statistical analysis")
            return
        
        print("--- Statistical Summary (Numerical Features) ---")
        display(self.df.describe())
        
        print("\n--- Correlation Matrix ---")
        # Selecting only numeric columns for correlation
        numeric_cols = self.df.select_dtypes(include=[np.number])
        corr_matrix = numeric_cols.corr()
        
        # Display correlation with Profit and Sales
        display(corr_matrix[['Sales', 'Profit']])
        return corr_matrix



# ---------------   Visualizations   ---------------------------
    def plot_sales_and_profit_by_category(self):
        """Plot total sales and profit by product category"""
        if self.df is None:
            print("No data available for plotting")
            return
        
        plt.figure(figsize=(10, 5))
        category_group = self.df.groupby('Category')[['Sales', 'Profit']].sum().reset_index()
        
        # Bar plot for Sales and Profit
        bar_width = 0.35
        x = np.arange(len(category_group['Category']))
        
        plt.bar(x - bar_width/2, category_group['Sales'], bar_width, label='Sales', color='royalblue')
        plt.bar(x + bar_width/2, category_group['Profit'], bar_width, label='Profit', color='seagreen')
        
        plt.xlabel('Category', fontsize=12)
        plt.ylabel('Amount', fontsize=12)
        plt.title('Total Sales and Profit by Category', fontsize=14, fontweight='bold')
        plt.xticks(x, category_group['Category'])
        plt.legend()
        plt.grid(axis='y', linestyle='--', alpha=0.7)
        plt.tight_layout()
        plt.show()

    def plot_profit_distribution(self):
        """Plot the distribution of profits using a histogram and KDE"""
        if self.df is None:
            print("No data available for plotting")
            return
        
        plt.figure(figsize=(10, 5))
        sns.histplot(self.df['Profit'], bins=50, kde=True, color='purple')
        plt.xlabel('Profit', fontsize=12)
        plt.ylabel('Frequency', fontsize=12)
        plt.title('Profit Distribution Analysis', fontsize=14, fontweight='bold')
        plt.xlim(-500, 1000)  # Zooming into the main distribution range for clarity
        plt.grid(True, linestyle='--', alpha=0.7)
        plt.tight_layout()
        plt.show()

    def plot_sales_by_region(self):
        """Plot total sales and profit by region."""
        if self.df is None: return
        plt.figure(figsize=(10, 4))
        region_df = self.df.groupby('Region')[['Sales', 'Profit']].sum().reset_index()
        x = np.arange(len(region_df['Region']))
        width = 0.35
        plt.bar(x - width/2, region_df['Sales'], width, label='Sales', color='darkorange')
        plt.bar(x + width/2, region_df['Profit'], width, label='Profit', color='forestgreen')
        plt.xticks(x, region_df['Region'])
        plt.title('Sales and Profit by Region')
        plt.legend()
        plt.grid(axis='y', linestyle='--', alpha=0.7)
        plt.tight_layout()
        plt.show()

    def plot_sales_by_segment(self):
        """Plot sales share by customer segment using a pie chart"""
        if self.df is None: return
        plt.figure(figsize=(6, 6))
        segment_df = self.df.groupby('Segment')['Sales'].sum()
        plt.pie(segment_df, labels=segment_df.index, autopct='%1.1f%%', startangle=140, colors=['skyblue', 'lightgreen', 'gold'])
        plt.title('Sales Share by Customer Segment')
        plt.tight_layout()
        plt.show()

    def plot_monthly_sales_trend(self):
        """Plot monthly sales trend over time"""
        if self.df is None: return
        plt.figure(figsize=(12, 4))
        monthly_sales = self.df.resample('ME', on='Order Date')['Sales'].sum()
        plt.plot(monthly_sales.index, monthly_sales.values, marker='o', color='crimson', linewidth=2)
        plt.title('Monthly Sales Trend Over Time')
        plt.xlabel('Order Date')
        plt.ylabel('Total Sales')
        plt.grid(True, linestyle='--', alpha=0.7)
        plt.tight_layout()
        plt.show()

    def plot_discount_vs_profit(self):
        """Scatter plot showing the impact of discount on profit"""
        if self.df is None: return
        plt.figure(figsize=(10, 4))
        sns.scatterplot(x='Discount', y='Profit', data=self.df, alpha=0.5, color='teal')
        plt.title('Impact of Discount on Profit')
        plt.grid(True, linestyle='--', alpha=0.7)
        plt.tight_layout()
        plt.show()

    def plot_top_sub_categories(self):
        """Plot top 10 sub-categories by total sales"""
        if self.df is None: return
        plt.figure(figsize=(10, 5))
        top_sub = self.df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=True).tail(10)
        top_sub.plot(kind='barh', color='slateblue')
        plt.title('Top 10 Sub-Categories by Sales')
        plt.xlabel('Total Sales')
        plt.grid(axis='x', linestyle='--', alpha=0.7)
        plt.tight_layout()
        plt.show()

    def plot_ship_mode_distribution(self):
        """Plot distribution of shipping modes"""
        if self.df is None: return
        plt.figure(figsize=(8, 4))
        sns.countplot(x='Ship Mode', data=self.df, order=self.df['Ship Mode'].value_counts().index, palette='Set2')
        plt.title('Distribution of Ship Modes')
        plt.xlabel('Ship Mode')
        plt.ylabel('Count')
        plt.grid(axis='y', linestyle='--', alpha=0.7)
        plt.tight_layout()
        plt.show()



    def generate_kpi_summary(self):
        """Generate automated KPI summaries"""
        if self.df is None: return
        print("\n" + "="*50)
        print("          AUTOMATED KPI ANALYTICAL REPORT          ")
        print("="*50)
        print(f"• Total Sales            : ${self.df['Sales'].sum():,.2f}")
        print(f"• Total Profit           : ${self.df['Profit'].sum():,.2f}")
        print(f"• Avg Profit Margin      : {self.df['Profit Margin'].mean():.2f}%")
        print(f"• Top Sales Category     : {self.df.groupby('Category')['Sales'].sum().idxmax()}")
        print("="*50 + "\n")

    def export_visual_reports(self):
        """Save a summary figure automatically"""
        if self.df is None: return
        plt.figure(figsize=(8, 4))
        self.df.groupby('Category')['Sales'].sum().plot(kind='bar', color='royalblue')
        plt.title('Category Sales Report')
        plt.savefig('Visual_Summary_Report.png', dpi=300)
        plt.close()
        print("Visual report exported as 'Visual_Summary_Report.png'")

    


# Initialize analyzer and process data
analyzer = SuperstoreAnalysis('Sample - Superstore 2019.xls')
analyzer.load_data()
analyzer.clean_and_engineer_features()

#optimize_memory
analyzer.optimize_memory_usage()
# Handle Outliers & Export Data 
analyzer.handle_outliers('Sales')


# Run  EDA checks
analyzer.perform_eda_checks()
print("="*50+"\n \n \n")
# Get statistical insights
analyzer.get_statistical_summary()



# Display all 8 visualizations
analyzer.plot_sales_and_profit_by_category()
analyzer.plot_profit_distribution()
analyzer.plot_sales_by_region()
analyzer.plot_sales_by_segment()
analyzer.plot_monthly_sales_trend()
analyzer.plot_discount_vs_profit()
analyzer.plot_top_sub_categories()
analyzer.plot_ship_mode_distribution()

# 4. Reporting & Exporting 
analyzer.generate_kpi_summary()
analyzer.export_cleaned_data('Cleaned_Superstore.csv')
analyzer.export_visual_reports()


```
